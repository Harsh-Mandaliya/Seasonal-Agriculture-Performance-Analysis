# Seasonal Agriculture Performance Analysis

This notebook analyzes seasonal differences in agricultural performance and develops evidence-based insights and recommendations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("data/agriculture_data.csv")
print("Shape:", df.shape)
df.head()
```

## 1. Dataset Understanding

```python
print(df.info())
display(df.describe(include="all").T)
print("Missing values:")
display(df.isna().sum().sort_values(ascending=False).head(15))
```

## 2. Seasonal Performance

```python
season_summary = df.groupby("Season").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Total_Production_Tonnes=("Production_Tonnes","sum"),
    Avg_Revenue_INR=("Revenue_INR","mean"),
    Avg_Cost_INR=("Total_Cost_INR","mean"),
    Avg_Profit_INR=("Profit_INR","mean"),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean")
).sort_values("Avg_Profit_INR", ascending=False)
display(season_summary.round(2))
```

```python
fig, ax = plt.subplots()
season_summary["Avg_Yield_Tonnes_Ha"].plot(kind="bar", ax=ax)
ax.set_title("Average Yield by Season")
ax.set_ylabel("Tonnes/Ha")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Crop Performance

```python
crop_summary = df.groupby("Crop").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Avg_Profit_INR=("Profit_INR","mean"),
    Total_Production_Tonnes=("Production_Tonnes","sum")
).sort_values("Avg_Profit_INR", ascending=False)
display(crop_summary.head(15).round(2))
```

## 4. Irrigation and Water Efficiency

```python
irrigation_summary = df.groupby("Irrigation_Method").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Avg_Water_Used_m3=("Water_Used_m3","mean"),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean"),
    Avg_Profit_INR=("Profit_INR","mean")
).sort_values("Avg_Water_Efficiency", ascending=False)
display(irrigation_summary.round(2))
```

## 5. Environmental and Risk Relationships

```python
corr = df.select_dtypes(include=np.number).corr(numeric_only=True)["Yield_Tonnes_Ha"].drop("Yield_Tonnes_Ha").sort_values(key=abs, ascending=False)
display(corr.to_frame("Correlation with Yield"))
```

```python
fig, ax = plt.subplots()
ax.scatter(df["Rainfall_mm"], df["Yield_Tonnes_Ha"], alpha=0.35)
ax.set_title("Rainfall vs Yield")
ax.set_xlabel("Rainfall (mm)")
ax.set_ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots()
ax.scatter(df["Disease_Pest_Risk_pct"], df["Yield_Tonnes_Ha"], alpha=0.35)
ax.set_title("Disease/Pest Risk vs Yield")
ax.set_xlabel("Disease/Pest Risk (%)")
ax.set_ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```
